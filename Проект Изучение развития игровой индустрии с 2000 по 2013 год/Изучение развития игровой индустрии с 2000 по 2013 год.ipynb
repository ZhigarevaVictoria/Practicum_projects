### Типы данных

Если встречаются некорректные типы данных, предположим их причины.
При необходимости проведем преобразование типов данных.

Некорректные типы данных в столбцах возникают из-за особенностей CSV-файлов. Pandas по умолчанию помечает столбцы как object, если встречает нечисловые значения вроде пустых строк, "N/A", буквенных оценок или пропусков.

В числовых столбцах могут встретиться строковые значения, например `unknown` или другие. Приводемтакие столбцы к числовому типу данных, заменив строковые значения на пропуски.

```python
# Изменяем тип данных
df['critic_score'] = pd.to_numeric(df['critic_score'], errors='coerce', downcast='float')
df['eu_sales'] = pd.to_numeric(df['eu_sales'], errors='coerce', downcast='float')
df['jp_sales'] = pd.to_numeric(df['jp_sales'], errors='coerce', downcast='float')
df['user_score'] = pd.to_numeric(df['user_score'], errors='coerce', downcast='float')
df['year_of_release'] = pd.to_datetime(df['year_of_release'], format='%Y', errors='coerce')
df['rating'] = df['rating'].astype('category')
df.info()
```

Изучим данные с пропущенными значениями. Сделаем промежуточный вывод: для каких столбцов характерны пропуски и сколько их. Предположим, почему пропуски могли возникнуть.


Работа с пропусками:
   - `name` и `genre`имеют меньше 5% пропусков, поэтому строки можно удалить без потери данных
   - `year_of_release`имеет 1,6% пропусков, заменяем пропуски в столбце на медиану для каждой платформу и жанра
   - `critic_score`, `user_score`, `rating`игнорируем пропуски, так как их слишком много (от 40 до 50%) — удаление значительно повлияет на данные, лучше провести анализ с индикаторами NaN

Предположим возможные причины пропусков:
   - `name`, вероятно, случайные ошибки или ошибка загрузки данных в датафрейм, так как пропусков всего 2
   - `year_of_release`возможно, в первоисточнике информации из интернета не был указан год релиза
   - `critic_score`, `user_score`, `rating` это критерии являются необязательными, поэтому эта информация может не добавляться (например, игра непопулярная, не набрала достаточно оценок и т.д.).

Обработаем пропущенные значения. Для каждого случая выберем оптимальный вариант: заменить на определённое значение, оставить как есть или удалить.

```python
# Удаляем пропуски в столбцах name и genre
df = df.dropna(subset=['name', 'genre'])
df.shape[0]
```

```python
# Заменяем пропуски в year_of_release на медиану для каждой платформу и жанра
def fill_year_median(row):
    if pd.isna(row['year_of_release']):
        # Находим строки с такой же комбинацией
        group = df[(df['platform'] == row['platform']) & 
                   (df['genre'] == row['genre'])]
        # Возвращаем медиану
        return group['year_of_release'].median()
    else:
        return row['year_of_release']

# Заполняем пропуски
df['year_of_release'] = df.apply(fill_year_median, axis=1)

# Удаляем строки, где year_of_release все еще NaN
df = df.dropna(subset=['year_of_release'])

print(f"Осталось строк после удаления: {df.shape[0]}")
```

## Фильтрация данных

Коллеги хотят изучить историю продаж игр в начале XXI века, и их интересует период с 2000 по 2013 год включительно. Отберите данные по этому показателю. Сохраните новый срез данных в отдельном датафрейме, например `df_actual`.

```python
# Отфильтруем данные по году выпуска игры с 2000 по 2013 год
df_actual = df[(df['year_of_release'] >= '2000-01-01') & (df['year_of_release'] < '2014-01-01')]
print(df_actual)
```

## Категоризация данных
    
Проведем категоризацию данных.
Разделим все игры по оценкам пользователей и выделим такие категории: высокая оценка (от 8 до 10 включительно), средняя оценка (от 3 до 8, не включая правую границу интервала) и низкая оценка (от 0 до 3, не включая правую границу интервала).

```python
# Категоризируем данные по пользовательской оценке
def categorize_user_score(score):
    if pd.isna(score):
        return 'Нет оценки'
    elif score >= 8:
        return 'Высокая оценка'
    elif score >= 3:
        return 'Средняя оценка'
    else:
        return 'Низкая оценка'

# Применяем категоризацию
df_actual['user_score_category'] = df_actual['user_score'].apply(categorize_user_score)

print(df_actual)
```

Разделим все игры по оценкам критиков и выделим такие категории: высокая оценка (от 80 до 100 включительно), средняя оценка (от 30 до 80, не включая правую границу интервала) и низкая оценка (от 0 до 30, не включая правую границу интервала).

```python
# Категоризируем данные по оценкам критиков
def categorize_critic_score(score):
    if pd.isna(score):
        return 'Нет оценки'
    elif score >= 80:
        return 'Высокая оценка'
    elif score >= 30:
        return 'Средняя оценка'
    else:
        return 'Низкая оценка'

# Применяем категоризацию
df_actual['critic_score_category'] = df_actual['critic_score'].apply(categorize_critic_score)

print(df_actual)
```

После категоризации данных проверим результат: сгруппируем данные по выделенным категориям и посчитаем количество игр в каждой категории.

```python
# Посчитаем количество строк для каждой категории `user_score_category`
count_user_score_category = df_actual['user_score_category'].value_counts()
print(count_user_score_category)
```

```python
# Посчитаем количество строк для каждой категории `critic_score_category`
count_critic_score_category = df_actual['critic_score_category'].value_counts()
print(count_critic_score_category)
```
